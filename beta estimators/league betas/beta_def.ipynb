```python
import pandas as pd
import joblib
import sys
from pathlib import Path

sys.path.append(str(Path.cwd().parent))

from utils  import fit_position_pipeline, select_features_by_correlation
```

```python
# features = joblib.load('../predictors/hist/feats_def')
features =  [
    'was_home', 'difficulty', 'opponent_difficulty', 'win_prob', 'elo_diff', 'yellow_cards_rolling_5', 'starts_rolling_3',  'points_3_ewm', 'xP_3_ewm',
    'minutes_rolling_3', 'goals_scored_3_ewm', 'assists_3_ewm', 'clean_sheets_3_ewm', 'goals_conceded_3_ewm', 'ict_index_3_ewm', 'expected_goals_3_ewm', 'expected_assists_3_ewm',
    'expected_goal_involvements_3_ewm', 'expected_goals_conceded_3_ewm', 'interceptions_3_ewm', 'blocks_3_ewm', 'clearances_3_ewm', 'tackles_3_ewm',
    'shots_faced_3_ewm',  'goals_prevented_3_ewm',
    ]

player_data = pd.read_csv('../../data/rolled_data_24_25.csv')
player_data = player_data[player_data['position'] == 'Defender']

league_models = {}

for gw in range(6,39):
    print(f"\n{'=' * 60}\n Model for round {gw} \n{'=' * 60}")
    data = player_data[(player_data['round']>3) & (player_data['round'] < gw)].fillna(0)
    feats = features #features[gw]
    # feats.remove('xP')
    selected_feats = select_features_by_correlation(data, feats, target_col='xP', redundancy_threshold=0.4, min_corr=0.2, verbose=True)


    league_model = fit_position_pipeline(
            data, selected_feats, gw, position_name='DEF', random_seed=42
        )

    print(f"Features for DEF kept in gw {gw} -> {league_model['feats']}")

    league_models[gw] = league_model
```

```python
joblib.dump(league_models, './league_models_def')
```
